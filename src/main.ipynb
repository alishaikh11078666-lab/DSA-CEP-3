**CEP–3: File System Directory Structure Simulation**

```python
import os

# 1. N-ary Tree Node
class FileNode:
    def __init__(self, name, is_dir=True):
        self.name = name
        self.is_dir = is_dir
        self.children = {}  # Hash Table for children (O(1) access)
        self.parent = None

class FileSystem:
    def __init__(self):
        self.root = FileNode("/")
        # 2. Stack for Path Traversal
        self.cwd_stack = [self.root]
        # 3. Hash Table for Global Search
        self.global_index = {"/": [self.root]}

    def create(self, name, is_dir=True):
        """Requirement: Create files and directories"""
        curr = self.cwd_stack[-1]
        if name in curr.children:
            print(f"Error: '{name}' already exists.")
            return

        new_node = FileNode(name, is_dir)
        new_node.parent = curr
        curr.children[name] = new_node

        # Add to Global Hash Table for Search
        if name not in self.global_index:
            self.global_index[name] = []
        self.global_index[name].append(new_node)
        print(f"{'Directory' if is_dir else 'File'} '{name}' created.")

    def delete(self, name):
        """Requirement: Delete files and directories"""
        curr = self.cwd_stack[-1]
        if name in curr.children:
            node_to_del = curr.children.pop(name)
            # Remove from Global Index
            self.global_index[name].remove(node_to_del)
            print(f"'{name}' deleted.")
        else:
            print(f"Error: '{name}' not found.")

    def search(self, name):
        """Requirement: Search files by name using Hash Table"""
        results = self.global_index.get(name, [])
        if not results:
            print(f"No results found for '{name}'.")
        for node in results:
            path = self._get_path(node)
            type_str = "DIR" if node.is_dir else "FILE"
            print(f"Found [{type_str}]: {path}")

    def _get_path(self, node):
        """Helper for Path Traversal using parent pointers"""
        path = []
        while node:
            path.append(node.name)
            node = node.parent
        return "/".join(reversed(path)).replace("//", "/")

    def display_hierarchy(self, node=None, indent=0):
        """Requirement: Display hierarchical structure (Recursive Tree Walk)"""
        if node is None: node = self.cwd_stack[-1]

        marker = "[D]" if node.is_dir else "[F]"
        print("  " * indent + f"{marker} {node.name}")
        for child in node.children.values():
            self.display_hierarchy(child, indent + 1)

# --- EXECUTION ---
fs = FileSystem()

# Step 1: Create Directories and Files
fs.create("home", is_dir=True)
fs.create("bin", is_dir=True)

# Step 2: Navigate and Create (Path Traversal)
home_node = fs.root.children["home"]
fs.cwd_stack.append(home_node) # cd home
fs.create("Qadeer", is_dir=True)
fs.cwd_stack.append(home_node.children["Qadeer"]) # cd Qadeer
fs.create("project.py", is_dir=False)

# Step 3: Search and Display
print("\n--- Global Search for 'project.py' ---")
fs.search("project.py")

print("\n--- Full Directory Hierarchy ---")
fs.display_hierarchy(fs.root)
```
